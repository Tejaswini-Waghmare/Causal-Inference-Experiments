# 03 — Randomized Experiments
Random assignment, balance, intent-to-treat, uncertainty

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('randomized_experiment.csv')
rates=df.groupby('group').converted.agg(['mean','sum','count']);print(rates)
from scipy.stats import norm
a=df.loc[df.group=='treatment','converted'];b=df.loc[df.group=='control','converted']
diff=a.mean()-b.mean();se=np.sqrt(a.var(ddof=1)/len(a)+b.var(ddof=1)/len(b));print('Difference and 95% Wald CI:',diff,(diff-1.96*se,diff+1.96*se))
print('Randomization identifies effects in expectation; interpret uncertainty.')
```
