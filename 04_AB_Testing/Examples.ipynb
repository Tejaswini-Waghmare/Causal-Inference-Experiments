# 04 — A/B Testing and Experiment Design
Metrics, power, guardrails, sample ratio mismatch

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('randomized_experiment.csv')
counts=df.group.value_counts();print('Allocation:',counts.to_dict())
from scipy.stats import chisquare
print('Sample-ratio mismatch test:',chisquare(counts.values))
print('Conversion:',df.groupby('group').converted.mean())
print('Revenue per visitor:',df.groupby('group').revenue.mean())
```
