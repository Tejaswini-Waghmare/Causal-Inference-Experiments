# 07 — Difference-in-Differences
Parallel trends, two periods and group-time comparisons

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('did_store_panel.csv')
means=df.groupby(['treated','post']).sales.mean().unstack();print(means)
did=(means.loc[1,1]-means.loc[1,0])-(means.loc[0,1]-means.loc[0,0]);print('DiD estimate:',did)
means.T.plot(marker='o',title='Pre/post trends (only one pre-period)');plt.show()
print('Parallel trends cannot be tested with a single pre-period.')
```
