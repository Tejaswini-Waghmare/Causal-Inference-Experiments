# 09 — Regression Discontinuity
Threshold designs, local effects and manipulation

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('rdd_scholarships.csv');bw=10
local=df[(df.score>=70-bw)&(df.score<=70+bw)].copy()
from sklearn.linear_model import LinearRegression
left=local[local.score<70];right=local[local.score>=70]
left_model=LinearRegression().fit(left[['score']],left.outcome);right_model=LinearRegression().fit(right[['score']],right.outcome)
jump=right_model.predict(pd.DataFrame({'score':[70]}))[0]-left_model.predict(pd.DataFrame({'score':[70]}))[0]
print('Local linear jump at cutoff:',jump,'Bandwidth:',bw)
plt.scatter(df.score,df.outcome,s=6,alpha=.3);plt.axvline(70,color='red');plt.show()
print('Assumes continuity of potential outcomes at threshold and no sorting.')
```
