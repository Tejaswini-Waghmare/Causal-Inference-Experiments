# 10 — End-to-End Causal Case Study
Campaign impact, diagnostics, sensitivity and reporting

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('campaign_observational.csv')
from sklearn.linear_model import LinearRegression
naive=df.groupby('treated').post_spend.mean().diff().iloc[-1]
reg=LinearRegression().fit(df[['treated','age','loyalty','prior_spend']],df.post_spend)
adjusted=reg.coef_[0]
print('Naive association:',round(naive,2),'Adjusted coefficient:',round(adjusted,2))
print('Group counts:',df.treated.value_counts().to_dict())
print('Pre-treatment balance:\n',df.groupby('treated')[['age','loyalty','prior_spend']].mean().round(2))
print('Recommendation: treat results as observational and assess unmeasured confounding.')
```
