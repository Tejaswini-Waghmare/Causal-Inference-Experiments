# 01 — Causal Thinking and DAGs
Confounding, counterfactuals, selection bias, and causal diagrams

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('campaign_observational.csv')
print(df.groupby('treated')[['prior_spend','loyalty','post_spend']].mean().round(2))
print('Naive observed difference:',df.groupby('treated').post_spend.mean().diff().iloc[-1])
# Treatment is not randomized; group differences are not necessarily causal.
df.boxplot(column='prior_spend',by='treated');plt.suptitle('');plt.title('Confounding check');plt.show()
```
