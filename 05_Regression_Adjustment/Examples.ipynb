# 05 — Regression Adjustment
Covariate adjustment and model limitations

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('campaign_observational.csv')
from sklearn.linear_model import LinearRegression
model=LinearRegression().fit(df[['treated','prior_spend','loyalty','age']],df.post_spend)
print('Adjusted treatment coefficient:',model.coef_[0])
print('Naive association:',df.groupby('treated').post_spend.mean().diff().iloc[-1])
print('Assumption: all important confounders measured and model specified appropriately.')
```
