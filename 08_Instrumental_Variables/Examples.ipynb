# 08 — Instrumental Variables
Relevance, exclusion restriction, Wald estimand

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('iv_encouragement.csv')
means=df.groupby('encouraged')[['participated','outcome']].mean();print(means)
first=means.loc[1,'participated']-means.loc[0,'participated'];reduced=means.loc[1,'outcome']-means.loc[0,'outcome']
print('First stage:',first,'Wald IV estimate:',reduced/first)
print('Requires relevance, exclusion, independence, monotonicity for LATE.')
```
