# 02 — Potential Outcomes and Treatment Effects
ATE, ATT, SUTVA, identification and estimands

**Data are synthetic.** Run from repository root or upload the CSV file to Colab.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def load_csv(name):
    candidates=[Path('data')/name, Path('../data')/name, Path('/content')/name]
    for p in candidates:
        if p.exists():
            return pd.read_csv(p)
    raise FileNotFoundError(f'Upload {name} in Colab, or run from repository root.')

df=load_csv('campaign_observational.csv')
# Simulation: two potential outcomes are not jointly observable in real data.
rng=np.random.default_rng(42); baseline=30+.67*df.prior_spend+15*df.loyalty+rng.normal(0,18,len(df))
y0=baseline;y1=baseline+9
print('Simulated true ATE:',(y1-y0).mean())
print('Observed association:',df.groupby('treated').post_spend.mean().diff().iloc[-1])
```
